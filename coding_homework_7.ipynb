```python
# I imported matplotlib and numpy

import matplotlib.pyplot as plt
import numpy as np
```

```python
# I used numpy in order to simplify the data sets
url_1 = 'https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/acc_hits_hw5.csv'
url_2 = 'https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/acc_fa_hw5.csv'

acc_hits = np.loadtxt(url_1, delimiter = ',')
acc_fa = np.loadtxt(url_2, delimiter = ',')
```

```python
# I used the shape function in order to find the number of trials and the number of samples for both acc_hits and acc_fa
print(acc_hits.shape)
print(acc_fa.shape)
```

```python
# I used the np.linspace syntax in order to create a time variable (both acc_hits and acc_fa have the same number of samples)
time = np.linspace(0, 4, 8001)
```

```python
# to find the sampling rate I divided the number of samples by the total length of each sample
sampling_rate = 8001 / 4
print(sampling_rate)
```

```python
# I used the ax syntax in order to create two subplots, one for acc_hits and one for acc_fa
fig, ax = plt.subplots(2,1, figsize = (15, 5))
ax[0].plot(time, acc_hits[0])
ax[0].set_ylabel('Voltage [\mu V]')

ax[1].plot(time, acc_fa[0])
ax[1].set_ylabel('Voltage [\mu V]')
ax[1].set_xlabel('Time (s)')
```

Since the clear LFP activity seemed to begin at the 2 second mark (when the stimulus was presented), the dominant oscillation for the acc_hits appears to be around 3Hz. As for the acc_fa, the dominant oscillation appears to be around 2Hz.

```python
# I used the code from our EPR notebook to plot the ERP of acc_hits and add a black vertical line at the two secnd mark when the screen touch occured.
plt.plot(time, acc_hits.mean(0))
plt.xlabel('Time [s]')
plt.ylabel('Voltage [$\mu V$]')
plt.title('ERP of acc_hits')

plt.plot([2, 2], [-45,45 ], 'k', lw=2)
plt.hlines(0, time[0], time[-1], color = 'black')
```

Yes, the appears to be consistant activity relation to the black bar which represents when the stimulus was presented. Before the presentation of the stimulus, there does not appear to be much consistent/ strong activity. After the stimulus was presented at 2 seconds, there is a sharp negative spike that then rebounds to a strong positive peak, seemingly displaying a full cycle.

```python
# I repeated what I did in the previous code cell but with acc_fa
plt.plot(time, acc_fa.mean(0))
plt.xlabel('Time [s]')
plt.ylabel('Voltage [$\mu V$]')
plt.title('ERP of acc_fa')

plt.plot([2, 2], [-10,10 ], 'k', lw=2)
plt.hlines(0, time[0], time[-1], color = 'black')
```

Yes there are clear differences between the the acc_hits plot and the acc_fa plot. For the hits ERP, there is a clear cycle, with a trough and a peak, right after the stimulus is presented. Comparatively, the false alarm plot does not seem to display any real consistant activity in relation to the presentation of the stimulus.
