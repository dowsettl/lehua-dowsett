```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
url_1 = 'https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/fc_hits.csv'
url_2 = 'https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/fc_mistakes.csv'
url_3 = 'https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/ic_hits.csv'
url_4 = 'https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/ic_mistakes.csv'

fc_hits = np.loadtxt(url_1, delimiter = ',')
fc_mistakes = np.loadtxt(url_2, delimiter = ',')
ic_hits = np.loadtxt(url_3, delimiter = ',')
ic_mistakes = np.loadtxt(url_4, delimiter = ',')
```

```python
print(fc_hits.shape)
print(fc_mistakes.shape)
print(ic_hits.shape)
print(ic_mistakes.shape)
```

```python
time = np.linspace(0, 4, 8001)
```

```python
fig, ax = plt.subplots(1,2,figsize=(15,5))
ax[0].plot(time,fc_hits.mean(0), color = 'blue')
ax[0].plot(time,fc_mistakes.mean(0), 'r')
ax[0].set_title('Frontal Cortex')
ax[0].set_ylabel('Voltage [\mu V]')
ax[0].set_xlabel('Time[s]')
ax[0].plot([2, 2], [-20,20], 'k', lw=2)
ax[0].hlines(0, time[0], time[-1], color = 'black')

ax[1].plot(time, ic_hits.mean(0), color = 'blue')
ax[1].plot(time, ic_mistakes.mean(0), 'r')
ax[1].set_title('Insular Cortex')
ax[1].set_ylabel('Voltage [\mu V]')
ax[1].set_xlabel('Time[s]')
ax[1].plot([2, 2], [-20,20], 'k', lw=2)
ax[1].hlines(0, time[0], time[-1], color = 'black')
```

Based on the hits and mistakes from the frontal cortex graph, the ERP appears to be larger with the hits than the mistakes. Similarly, in the graph displaying the insular cortex data, the ERP also appears to be larger with the hits than the mistakes. In both the frontal cortex and the insular cortex, immediately after the mouse touched the screen, there was a large negative spike followed by a large positive spike with the hits. As for the mistakes, there seemed to be a lot of noise, with no significant activity in relation to the mouse touching the screen.
