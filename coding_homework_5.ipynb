```python
# I imported matplotlib and numpy

from matplotlib import pyplot as plt
import numpy as np
```

```python
# I used numpy to simplify the spike data into arrays.
url_1 = 'https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/spikes_left_trial1.csv'
url_2 = 'https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/spikes_right_trial1.csv'
url_3 = 'https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/spikes_left_trial2.csv'
url_4 = 'https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/spikes_right_trial2.csv'

spikes_left_trial1 = np.loadtxt(url_1, delimiter = ',')
spikes_right_trial1 = np.loadtxt(url_2, delimiter = ',')
spikes_left_trial2 = np.loadtxt(url_3, delimiter = ',')
spikes_right_trial2 = np.loadtxt(url_4, delimiter = ',')
```

```python
# Used the len() function to find the length of each array.
len_slt1 = len(spikes_left_trial1)
print(len_slt1)
len_srt1 = len(spikes_right_trial1)
print(len_srt1)
len_slt2 = len(spikes_left_trial2)
print(len_slt2)
len_srt2 = len(spikes_right_trial2)
print(len_srt2)
```

```python
# In order to get the firing rate for each trial, I divided the length of the array which is the number of spikes, and divided it by 30 because that is the length of the waiting period.
rate_slt1 = len_slt1 / 30
print(rate_slt1)
rate_srt1 = len_srt1 / 30
print(rate_srt1)
rate_slt2 = len_slt2 / 30
print(rate_slt2)
rate_srt2 = len_srt2 / 30
print(rate_srt2)
```

Based on the firing rates for both trials, it appears as though the neuron fires at fairly similar rates for both left and right turn trials. However, the average firing rate for left turns is slightly higher than that of the right turn trial.

```python
# I used the subplot function from matplot lib to create two separate graphs.
fig, ax = plt.subplots(2,1, figsize = (10,8))
ax[0].scatter(spikes_left_trial1, np.ones_like(spikes_left_trial1), marker = '.')
ax[0].scatter(spikes_right_trial1, 2 * np.ones_like(spikes_right_trial1), marker = '.')
ax[0].set_title('Trial 1')
ax[0].set_xlabel('Time [s]')
ax[0].set_ylim([0,3])
ax[0].set_yticks([1,2])
ax[0].set_yticklabels(['Left','Right'])



ax[1].scatter(spikes_left_trial2, np.ones_like(spikes_left_trial2), marker = '.')
ax[1].scatter(spikes_right_trial2, 2 * np.ones_like(spikes_right_trial2), marker = '.')
ax[1].set_title('Trial 2')
ax[1].set_xlabel('Time[s]')
ax[1].set_ylim([0,3])
ax[1].set_yticks([1,2])
ax[1].set_yticklabels(['Left','Right'])

plt.show()
```

Based on the two subplots, it appears as though the neuron is slightly more "bursty" with right turn trials. The spikes for the right turn appears to cluster more into groups, whereas the spikes for the left turns appear to be a little more spread out.

```python
# I used np.diff() function to compute the ISI for each spike array.
ISI_slt1 = np.diff(spikes_left_trial1)
print(ISI_slt1)
ISI_srt1 = np.diff(spikes_right_trial1)
print(ISI_srt1)
ISI_slt2 = np.diff(spikes_left_trial2)
print(ISI_slt2)
ISI_srt2 = np.diff(spikes_right_trial2)
print(ISI_srt2)
```

```python
# I made a bins variable and then used the fig, ax syntax to create four different histogram subplots.
bins = np.arange(0, 2.01, 0.01)
fig, ax = plt.subplots(2, 2, figsize=(12, 10))

ax[0, 0].hist(ISI_slt1, bins, color='blue')
ax[0, 0].set_title('ISI: Trial 1 - Left Turn')
ax[0, 0].set_ylabel('Count')

ax[0, 1].hist(ISI_srt1, bins, color='orange')
ax[0, 1].set_title('ISI: Trial 1 - Right Turn')

ax[1, 0].hist(ISI_slt2, bins, color='blue')
ax[1, 0].set_title('ISI: Trial 2 - Left Turn')
ax[1, 0].set_xlabel('Inter-Spike Interval (s)')
ax[1, 0].set_ylabel('Count')

ax[1, 1].hist(ISI_srt2, bins, color='orange')
ax[1, 1].set_title('ISI: Trial 2 - Right Turn')
ax[1, 1].set_xlabel('Inter-Spike Interval (s)')

plt.show()
```

The neuron does tend to firing in bursts with a strong concentration of short ISIs, around 0.0-0.25s, and then there is a long tail where they are more spread out, inidcating periods of silence between bursts. It appears as though the neuron has a higher burstiness during the right turn trials as compared to the left turn trials, with a higher peak at very short ISIs.

```python
# created a new bin variable and used the np.histogram function to get the firing rate over time for each trial. I then used the fig, ax function to create two subplots.
time_bins = np.arange(0, 31, 1)

fr_slt1 = np.histogram(spikes_left_trial1, time_bins)[0]
fr_srt1 = np.histogram(spikes_right_trial1, time_bins)[0]
fr_slt2 = np.histogram(spikes_left_trial2, time_bins)[0]
fr_srt2 = np.histogram(spikes_right_trial2, time_bins)[0]

fig, ax = plt.subplots(2, 1, figsize=(10, 8))
ax[0].plot(time_bins[:-1] + 0.5, fr_slt1, color='red', label='Left Turn')
ax[0].plot(time_bins[:-1] + 0.5, fr_srt1, color='blue', label='Right Turn')
ax[0].set_title('Firing Rate Across Time: Trial 1')
ax[0].set_ylabel('Firing Rate (Hz)')
ax[0].legend()

ax[1].plot(time_bins[:-1] + 0.5, fr_slt2, color='red', label='Left Turn')
ax[1].plot(time_bins[:-1] + 0.5, fr_srt2, color='blue', label='Right Turn')
ax[1].set_title('Firing Rate Across Time: Trial 2')
ax[1].set_xlabel('Time (s)')
ax[1].set_ylabel('Firing Rate (Hz)')
ax[1].legend()

plt.show()
```

There are differences between the firing rates at certain timepoints and the pattern differs between trial 1 and trial 2. In trial 1, the left turn shows more pronounced peaks are 12-16s whereas the right turn is generally lower in the middle of the 30s time period but shows a strong spike around 30s. In trial 2, the right turn has a strong spike early on, around 3s, as well as additional bursts around 9-15s whereas the left turn is relatively quite initially, but has more spikes between 21s and 30s.
